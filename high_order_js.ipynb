```python
%%writefile code.js

// Part A
const data = [5, 12, 3, 20, 11, 9, 30];

const loop = []
for (let i = 0; i<data.length; i++) {
  if (data[i] > 10) {
    loop.push(data[i]*2);
  }
}
console.log(loop);

const pipeline = data.filter(x => x > 10).map(x => x * 2);
console.log(pipeline);

// Part B
const numbers = [1, 2, 3, 4, 5, 6];
const reduce = numbers.filter(x => x % 2 == 0).map(x => x * 2).reduce((sum, n) => sum + n, 0);

console.log(reduce);
// Part C

const users = [
  { id: 1, name: "Alice", age: 25 },
  { id: 2, name: "Bob", age: 17 },
  { id: 3, name: "Carol", age: 19 }
];

const under18 = users.some((user) => {
  return user.age < 18
}, users)

const user3 = users.find((user) => {
  return user.id == 3
})

console.log(under18)
console.log(user3)
users.forEach((user) => {
  console.log(user.name)
})

// Part D

function repeat(n, action) {
  for (let i = 0; i < n; i++) {
    action(i);
  }
}

repeat(3, console.log);
repeat(4, i => console.log("Step "+i))








```
